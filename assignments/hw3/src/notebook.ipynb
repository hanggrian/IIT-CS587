> 3.  Calculate the overall defect removal effectiveness for the process

Using the combined defects value, we can compare it to the amount of total defects to determine the Defect Removal Effectiveness score.

$$
\begin{align}
  R \ \textsf{DRE} &= \left( 1 - \frac{R \ \textsf{defects}}{\textsf{Total defects}} \right) \times 100\% \\
  &= \left( 1 - \frac{52}{1054} \right) \times 100\% &= \mathbf{95.06\%} \\
  A \ \textsf{DRE} &= \left( 1 - \frac{A \ \textsf{defects}}{\textsf{Total defects}} \right) \times 100\% \\
  &= \left( 1 - \frac{49}{1054} \right) \times 100\% &= \mathbf{95.35\%} \\
  I0 \ \textsf{DRE} &= \left( 1 - \frac{I0 \ \textsf{defects}}{\textsf{Total defects}} \right) \times 100\% \\

  I1 \ \textsf{DRE} &= \left( 1 - \frac{I1 \ \textsf{defects}}{\textsf{Total defects}} \right) \times 100\% \\
  &= \left( 1 - \frac{299}{1054} \right) \times 100\% &= \mathbf{71.63\%} \\
  UT \ \textsf{DRE} &= \left( 1 - \frac{UT \ \textsf{defects}}{\textsf{Total defects}} \right) \times 100\% \\
  &= \left( 1 - \frac{246}{1054} \right) \times 100\% &= \mathbf{23.33\%} \\
  IT \ \textsf{DRE} &= \left( 1 - \frac{IT \ \textsf{defects}}{\textsf{Total defects}} \right) \times 100\% \\
  &= \left( 1 - \frac{114}{1054} \right) \times 100\% &= \mathbf{86.33\%} \\

> 4.  Create a stacked bar chart to show the Defects Originated, Detected, and Escaped for every phase.

First, calculate the required metrics using NumPy. Then, create a Pandas table data with pivoted column using `melt` function. Finally, use Seaborn to plot a bar chart using the table data.

````python
from matplotlib import pyplot
from numpy import diag
from pandas import DataFrame
from seaborn import barplot

originated = diag(frame)
detected = frame.sum(axis=1) - originated
escaped = [frame.iloc[i, i + 1:].sum() for i in range(len(frame))]

pyplot.figure(figsize=(12, 6))
barplot(
    data= \
        DataFrame({
            'Phase': frame.index,
            'Originated defects': originated,
            'Detected defects': detected,
            'Escaped defects': escaped,
        }).melt(
            id_vars='Phase',
            value_vars=[
                'Originated defects',
                'Detected defects',
                'Escaped defects',
            ],
            var_name='Metric',
            value_name='Defects',
        ),
    x='Phase', 
    y='Defects',
    hue='Metric',
)

pyplot.title('Defects Chart')
pyplot.xlabel('Phase')
pyplot.ylabel('Number of Defects')
pyplot.xticks(rotation=16) # text won't fit without rotation
pyplot.show()
````

> Implement the requirements in Part I using Python, Langchain/Langgraph and OpenAI/gpt-4-mini. The AI Agent representing the Software Project Manager shall use GenAI model to generate Python code and use Python REPL to execute the generated Python code.'

Supply the LLM configuration with API keys stored in environment variables.

````python
from os import getenv
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())

llm_config = {
    'model': 'gpt-4o-mini',
    'api_key': getenv('OPENAI_API_KEY'),
}
````

Describe the task, embedding the content from the CSV file.

````python
from textwrap import dedent

BOLD = '\033[1m'
RESET = '\033[0m'

# rename cells of first column to make it easier for agents
renamed_frame = \
    frame.rename(
        index={
            'Requirement (R)': 'R',
            'Analysis (A)': 'A',
            'Design (I0)': 'I0',
            'Coding (I1)': 'I1',
            'Unit Testing (UT)': 'UT',
            'Integration Testing (IT)': 'IT',
            'System Testing (ST)': 'ST',
            'Field (F)': 'F',
        },
    )

task = \
    'Generate a Python script that automates the evaluation of the defect removal effectiveness for the project below. ' + \
    'The script must be runnable in a Python REPL interactive shell session.\n\n' + \
    f'{renamed_frame}\n\n'

# insert sample code to improve accuracy
task += \
    'The formula for the defect removal effectiveness from IBM Houston is well-documented. ' + \
    'Consider the sample code below to and solve the task above using comparable technique:\n\n' + \
    dedent(
        """\
        ```python
        dre = {}
        for phase in matrix.index:
            detected = matrix.loc[phase, phase]
            escaped = matrix.loc[phase, phase+1:].sum() if phase != matrix.columns[-1] else 0
            dre[phase] = detected / (detected + escaped) if (detected + escaped) > 0 else None
        print(Series(dre))
        ```
        """,
    )

print(f'{BOLD}Description:{RESET}\n')
print(task)
````

Define customer and project manager agent as the main participants of the conversation. When the project manager receives the task, it triggers a nested conversation between inner agents to solve the task.

````python
from autogen import AssistantAgent, UserProxyAgent, GroupChat, GroupChatManager

cp = \
    UserProxyAgent(
        name='CP',
        description='Customer Proxy',
        human_input_mode='NEVER',
        is_termination_msg=lambda x: x.get('content', '').find('TERMINATE') >= 0,
        code_execution_config={
            'last_n_messages': 1,
            'work_dir': 'tasks',
            'use_docker': False,
        },
    )
pm = \
    AssistantAgent(
        name='PM',
        description='Project Manager',
        llm_config=llm_config,
        system_message= \
            'You are a project manager, a natural leader who excels in team organization. ' + \
            'You monitor project progress, set deadlines and manage the budget. ' + \
            'You take immediate action to overcome obstacles using your creative problem-solving skills.',
    )

inner_assistant = \
    AssistantAgent(
        'Inner-assistant',
        llm_config=llm_config,
        is_termination_msg=lambda x: x.get('content', '').find('TERMINATE') >= 0,
    )
inner_code_interpreter = \
    UserProxyAgent(
        'Inner-code-interpreter',
        human_input_mode='NEVER',
        code_execution_config={
            'work_dir': 'coding',
            'use_docker': False,
        },
        default_auto_reply='',
        is_termination_msg=lambda x: x.get('content', '').find('TERMINATE') >= 0,
    )

chat_manager = \
    GroupChatManager(
        groupchat= \
            GroupChat(
                agents=[inner_assistant, inner_code_interpreter],
                messages=[],
                speaker_selection_method='round_robin',
                allow_repeat_speaker=False,
                max_round=8,
            ),
        is_termination_msg=lambda x: x.get('content', '').find('TERMINATE') >= 0,
        llm_config=llm_config,
        code_execution_config={
            'work_dir': 'coding',
            'use_docker': False,
        },
    )
````

Start the chat!

````python
pm.register_nested_chats(
    [
        {
            'recipient': chat_manager,
            'summary_method': 'reflection_with_llm',
            'clear_history': False,
        },
    ],
    trigger=cp,
)
cp.initiate_chats([
    {
        'recipient': pm,
        'message': task,
        'max_turns': 1,
        'summary_method': 'last_msg',
    }
])
````
